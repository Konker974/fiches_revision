```python
from sqlalchemy import delete, func, insert, select
from sqlalchemy.orm import Session
import pandas as pd
from pathlib import Path
import os
```

```python
from create_db import get_engine, Personnal, Company
```

```python
engine = get_engine(os.environ.get(key='DATABASE_URL'))
```

```python
df= pd.read_csv('fin_exploration.csv')
```

```python
df = df.drop(columns=df.columns[0])
```

```python
COL_PERSONNES= [
"personal.first_name",
"personal.last_name",
"personal.email",
"personal.country",
"personal.age",
"personal.gender",
"personal.marital_status",
"job.salary",
"job.department",
"job.join_date",
"job.position_level",
"job.performance_score",
"job.last_promotion_date",
"experience.experience_years",
"experience.has_certifications",
"experience.education_level",
"experience.work_hours_per_week",
"experience.projects_completed",
"company.name"
]
```

```python
COL_COMPANIES= [
"company.name",
"company.num_employees",
"company.benefits",
"company.main_area_of_expertise",
"company.headquarters"
]
```

```python
COMPANIES = df['company.name'].unique().tolist()
```

```python
df.loc[0,COL_PERSONNES]
```

```python

with Session(engine) as session :
    row = df.iloc[0]
    company = Company(
            company_name=row['company.name'],
            company_num_employees=row['company.num_employees'],
            company_benefits=row['company.benefits'],
            company_main_area_of_expertise=row['company.main_area_of_expertise'],
            company_headquarters=row['company.headquarters'],
        )
    person = Personnal(
        personal_first_name= row['personal.first_name'],
        personal_last_name= row['personal.last_name'],
        personal_email= row['personal.email'],
        personal_country= row['personal.country'],
        personal_age= row['personal.age'],
        personal_gender= row['personal.gender'],
        personal_marital_status= row['personal.marital_status'],
        job_salary= row['job.salary'],
        job_department= row['job.department'],
        job_join_date= row['job.join_date'],
        job_position_level= row['job.position_level'],
        job_performance_score= row['job.performance_score'],
        job_last_promotion_date= row['job.last_promotion_date'],
        experience_experience_years= row['experience.experience_years'],
        experience_has_certifications= row['experience.has_certifications'],
        experience_education_level= row['experience.education_level'],
        experience_work_hours_per_week= row['experience.work_hours_per_week'],
        experience_projects_completed= row['experience.projects_completed'],
        company= company
    )
    session.add(person)
    session.commit()
```

```python
QUERY = """
SELECT *
FROM personnal
"""
```
